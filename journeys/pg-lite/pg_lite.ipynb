# Postgres in your browser, step by step

This journey is different: Postgres runs client-side using PGlite, a WebAssembly build of Postgres. There is no server connection and no Python script — instead, you open an HTML file in your browser and type SQL.

**Before you start:** finish `00-setup/`.

## What is PGlite?

PGlite is Postgres compiled to WebAssembly (WASM). It runs entirely in your browser — no server, no network, no latency. Same Postgres SQL semantics, same data types, same functions. It is real Postgres, shrunk to 10 MB and embedded in a web page.

## Step 1 — Open the interactive editor

Open the file `../../common/pglite-demo.html` in your browser (or save it and open it as file://). Right-click on the file and select 'Open with Browser', or double-click it.

You should see: 'PGlite ready. Type SQL and hit Run.'

This notebook explains the concept. The interactive editor is in that HTML file.

## Step 2 — Create a table

In the HTML editor's SQL pane, type or paste:

```sql
CREATE TABLE machines (
    id TEXT PRIMARY KEY,
    status TEXT,
    health NUMERIC
);
```

Hit 'Run SQL'. The table is created in the browser's memory.

## Step 3 — Insert data

```sql
INSERT INTO machines VALUES
    ('M-204', 'healthy', 91.5),
    ('M-205', 'derated', 72.0),
    ('M-206', 'maintenance', 45.0);
```

## Step 4 — Query the data

```sql
SELECT * FROM machines WHERE health > 70;
```

Results appear instantly in the browser. No server round-trip.

## Step 5 — Try window functions

```sql
SELECT
    id,
    status,
    health,
    ROW_NUMBER() OVER (ORDER BY health DESC) as rank
FROM machines
ORDER BY rank;
```

## Step 6 — Try JSONB aggregation

```sql
SELECT
    status,
    jsonb_agg(jsonb_build_object('id', id, 'health', health)) as machines
FROM machines
GROUP BY status;
```

## Step 7 — Try upsert (ON CONFLICT)

```sql
INSERT INTO machines VALUES ('M-204', 'operational', 99.0)
ON CONFLICT (id) DO UPDATE SET health = 99.0;
```

## Step 8 — Prove it is real Postgres

```sql
SELECT version();
```

You will see 'PostgreSQL on wasm32'. Same Postgres kernel.

## What you learned

Postgres runs client-side in WebAssembly. No server connection needed. Same SQL, same semantics, same reliability. This is how lightweight, embedded, and offline-first applications can use real Postgres — in the browser, on the phone, on the edge.
